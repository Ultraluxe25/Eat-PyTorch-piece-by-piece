### Fine-Tuning - это дообучение модели на более маленькой обучающей выборке из специфической области, чтобы сделать итоговую модель более "умной" в отдельной области.

Блок слоев **Self-Attention** позволяет модели оценивать важность слов и токенов в последовательности по отношению друг к другу.

### Блок кодировщика может заполнять пропуски (маски) и классифицировать текст. Блок декодирощика отвечает за генерацию.
